# Reign distributions
Reproduce the atlas summaries from the generated Parquet data. Exclusions use the published eligibility policy.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd() if Path("public/data").exists() else Path.cwd().parent
reigns = pd.read_parquet(root / "public/data/mart_ruler_reigns.parquet")
eligible = reigns[reigns.eligible]
eligible.groupby("realm_name").reign_years.agg(["count", "mean", "median", "min", "max"])
```

```python
# Curation coverage must accompany interpretation.
reigns.groupby("realm_name").agg(tenures=("reign_id", "count"), life_dates=("lifespan", "count"))
```
